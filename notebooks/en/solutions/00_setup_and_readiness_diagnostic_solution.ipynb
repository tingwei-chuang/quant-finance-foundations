**Q3. What does it mean that "the sample mean is a random variable"?**
- A. It gives a different result every time you compute it
- B. It depends on the sample drawn — a different sample gives a different value
- C. It cannot be computed
- D. It is always inaccurate

**Q4. What is the main problem with a random train/test split for time series?**
- A. It is too slow to compute
- B. The training set can contain "future" information, causing look-ahead bias
- C. The test set is too small
- D. You cannot make plots

```python
my_answers = {1: 'A', 2: 'B', 3: 'B', 4: 'B'}

import hashlib as _hashlib
_expected = {1: '4c241b5f35e88635', 2: '5b7f59eb090c222e', 3: '6e8e3e969a32d31f', 4: 'dc3227e7a70d29b8'}
_n_correct = 0
for _q, _ans in my_answers.items():
    if _ans is None:
        print(f'Q{_q}: unanswered')
        continue
    _h = _hashlib.sha256(f'qmr-w0-q{_q}-{str(_ans).strip().upper()}'.encode()).hexdigest()[:16]
    _ok = _h == _expected[_q]
    _n_correct += int(_ok)
    print(f'Q{_q}: ' + ('✔ correct' if _ok else '✘ incorrect'))
print(f'Score: {_n_correct} / {len(my_answers)}')
```

### Explanations

- **Q1 → A**: simple = 110/100 − 1 = 10%; log = ln(1.1) ≈ 9.53%. The larger the move, the bigger the gap between the two.
- **Q2 → B**: Synthetic data makes every result fully reproducible without network access, and avoids third-party data terms of use.
- **Q3 → B**: The sample mean is a function of the data; the data are the result of random sampling, so the sample mean itself has a sampling distribution.
- **Q4 → B**: A random split lets the model learn from future data and then "predict" the past, making results systematically over-optimistic.

## Common mistakes

- **Mixing up simple returns and log returns without realizing it.**
- **Using random data without a fixed seed, so results cannot be reproduced.**
- **Believing a strategy works just because its equity curve goes up.**
