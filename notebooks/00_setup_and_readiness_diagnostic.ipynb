## 小概念題解答

1. simple return = 110/100 − 1 = **0.10（10%）**；log return = ln(110/100) ≈ **0.0953**。漲幅越大，兩者差越多。

3. 因為隨機切分會讓模型用到「未來」的資料來預測「過去」，造成 look-ahead bias，使回測結果過度樂觀（Week 7、8 主題）。
